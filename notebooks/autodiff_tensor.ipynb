```python
import numpy as np
```

```python
def gradcheck(f, x, h=1e-6):
    for t in x:
        t.grad.fill(0)

    out = f(*x)
    out.backward()
    analytic = [t.grad.copy() for t in x]
    numeric = []

    for t in x:
        n = np.zeros_like(t.data)
        for idx in np.ndindex(t.data.shape):
            original = t.data[idx]
            t.data[idx] = original + h
            plus = f(*x).data
            t.data[idx] = original - h
            minus = f(*x).data
            t.data[idx] = original
            n[idx] = (plus - minus) / (2 * h)
        numeric.append(n)

    for a, n in zip(analytic, numeric):
        err = np.abs(a - n) / np.maximum(
            np.abs(a) + np.abs(n), 1e-12
        )
        print(err.max())
```
